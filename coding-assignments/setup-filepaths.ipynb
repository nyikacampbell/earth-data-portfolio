# Setup script to create file path and data directories to use in all assignments

```python
### setup
### folder/file management
import os
import pathlib
```

```python
### create overall portfolio directory
portfolio_dir = os.path.join(

    ### point to your home directory
    pathlib.Path.home(),

    ### nest the data in the folder where your assignments live
    'Repos',
    'earth-data-portfolio',
    )

### make the directory
os.makedirs(portfolio_dir, exist_ok = True)
portfolio_dir
```

```python
### create directory for the coding-assignments
assignments_dir = os.path.join(
    portfolio_dir,
    'coding-assignments'
    )

### make the directory
os.makedirs(assignments_dir, exist_ok = True)
assignments_dir
```

```python
### create image directory
img_dir = os.path.join(
    portfolio_dir,
    'img'
    )

### make the directory
os.makedirs(img_dir, exist_ok = True)
img_dir
```

```python
### create image directory
posts_dir = os.path.join(
    portfolio_dir,
    'portfolio-posts'
    )

### make the directory
os.makedirs(posts_dir, exist_ok = True)
posts_dir
```
