# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?
2. How might climate change affect animal migrations?
3. Do you notice any animal migrations where you live?

**Your answer here:**

1. Many animals migrate to find conditions that are more suitable, whether this is migrating away from cold weather or predation, toward food sources, or to places with conducive conditions to mate or raise young. Migrations may occur at any spatial or temporal scale ranging from the daily microscopic migrations of *Daphnia*  to the multi-thousand-mile flights of arctic turns (Chapman et al., 2014). Some animals like monarchs migrate over as many as 4 generations following seasons despite short lifespans (Alerstam & Bäckman, 2018). Other species such as American Robin have both resident and migrating populations. Migration is a relatively plastic trait and can arise or be suppressed over only a few generations in partially migratory populations (Alerstam & Bäckman, 2018). Interestingly, a recent metanalysis by Buchan et al. (2020) found that for birds, reptiles, and amphibians *resident* or non-migratory populations experience a fitness advantage largely driven by higher survival rates. This phenomenon is also mentioned anecdotally in the book [Wild ones](https://www.goodreads.com/en/book/show/16158510-wild-ones) by Jon Mooallem where Mooallem discusses a whooping crane population which settled permanently at a dump due to abundant food.

2. Animals use many different cues to trigger migration from temperature to daylength to decadal oscillations (Shaw, 2016). Some of these factors are impacted by climate change while others remain less affected. Additionally some migration disruptions can be tolerated, while others—specifically those to breeding migrations—cannot (Robinson et al., 2009). This makes it difficult to predict changes to migration globally. Bird migration is relatively well studied, particularly in the Northern hemisphere, where warming temperatures are favoring shorter or partial migration for species like blackcaps (*Sylvia atricapilla*) (Shaw, 2016). This can lead to overlap with other migrating populations increasing competition. Additionally, as Bauer and Hoye, (2014) point out migration is a significant driver of ecosystem function, spreading nutrients, carbon, and biomass to otherwise disconnected systems. Because of this climate-driven changes to migration have the potential to impact global ecosystem functions. The complexity of migration and the many channels through which migration changes will act make conservation efforts for migratory species particularly difficult (Robinson et al., 2009). Meaningful protection of migratory systems and the services they provide will require dedicated and coordinated efforts from governments, scientists, and citizens.

3. Growing up in Madison, Wisconsin we would see many seasonal bird migrations. The 1996 family movie [Fly Away Home]( https://en.wikipedia.org/w/index.php?title=Fly_Away_Home&oldid=1380502632) tells the story of a Dad and daughter who rescue orphaned geese and build an ultralight plane to teach them to migrate. This movie was my personal introduction to bird migration and restoration techniques. In Madison migrating geese were a common sight, but we sometimes saw cranes! In the fall my family would go to a nearby state park to walk their boardwalk trails and look for cranes. We were always on the lookout for endangered whooping cranes. I learned from a young age about the incredible efforts to reintroduce Whooping cranes, and their traditional migratory path, to the Midwest. Whooping cranes typically migrate from upper Midwest in summer down to the Southeast in the winter (Thompson et al., 2022). This migration was fully halted due to habitat loss and near-fatal population collapse in the 1940s when the population was reduced to 21 individuals across two fragment populations (Urbanek et al., 2010). In the 1960s dedicated and creative reintroduction efforts began using costumed caretakers and ultralight aircraft to socialize captive-breed cranes and teach the traditional migration path. These, along with other efforts, have brought the population to 76 individuals in 2021, 16 of which were born in the wild (Thompson et al., 2022).
